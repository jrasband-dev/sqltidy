```python
"""Test the updated Python API"""
import sqltidy

# Test SQL
sql = """
select name, email, (select count(*) from orders where orders.user_id = users.id) as order_count
from users where active = 1
"""

print(sql)
```

```python

# Test 1: tidy_sql with PostgreSQL
print("\n" + "=" * 70)
print("TEST 1: tidy_sql(dialect='postgresql')")
print("=" * 70)
result = sqltidy.tidy_sql(sql, dialect='postgresql')
print(result)
```

```python

# Test 2: tidy_sql with SQL Server (default)
print("\n" + "=" * 70)
print("TEST 2: tidy_sql(dialect='sqlserver')")
print("=" * 70)
result = sqltidy.tidy_sql(sql, dialect='sqlserver')
print(result)
```

```python
# Test 3: rewrite_sql (should convert subquery to CTE if enabled)
print("\n" + "=" * 70)
print("TEST 3: rewrite_sql()")
print("=" * 70)
result = sqltidy.rewrite_sql(sql)
print(result)
```

```python
# Test 4: tidy_and_rewrite_sql
print("\n" + "=" * 70)
print("TEST 4: tidy_and_rewrite_sql(dialect='postgresql')")
print("=" * 70)
result = sqltidy.tidy_and_rewrite_sql(sql, dialect='postgresql')
print(result)
```
