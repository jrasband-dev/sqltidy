```python
# Add the local sqltidy to the path to use the development version
import sys
from pathlib import Path

# Add the project root to sys.path
project_root = Path.cwd()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

# Force reload of sqltidy modules if already imported
if 'sqltidy' in sys.modules:
    import importlib
    for module_name in list(sys.modules.keys()):
        if module_name.startswith('sqltidy'):
            del sys.modules[module_name]

print(f"Using local sqltidy from: {project_root}")
```

```python
"""Test the updated Python API"""
import sqltidy

# Test SQL
sql = """
select name, email, (select count(*) from orders where orders.user_id = users.id) as order_count
from users where active = 1
"""

print(sql)
```

```python

# Test 1: tidy_sql with PostgreSQL
print("\n" + "=" * 70)
print("TEST 1: tidy_sql(dialect='postgresql')")
print("=" * 70)
result = sqltidy.tidy_sql(sql, dialect='postgresql')
print(result)
```

```python

# Test 2: tidy_sql with SQL Server (default)
print("\n" + "=" * 70)
print("TEST 2: tidy_sql(dialect='sqlserver')")
print("=" * 70)
result = sqltidy.tidy_sql(sql, dialect='sqlserver')
print(result)
```

```python
# Test 3: rewrite_sql (should convert subquery to CTE if enabled)
print("\n" + "=" * 70)
print("TEST 3: rewrite_sql()")
print("=" * 70)
result = sqltidy.rewrite_sql(sql)
print(result)
```

```python
# Test 4: tidy_and_rewrite_sql
print("\n" + "=" * 70)
print("TEST 4: tidy_and_rewrite_sql(dialect='postgresql')")
print("=" * 70)
result = sqltidy.tidy_and_rewrite_sql(sql, dialect='postgresql')
print(result)
```
